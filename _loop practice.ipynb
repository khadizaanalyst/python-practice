```python
a = "hello"
b = "123Hello"
c = "123456"
d = "HELLO"
e = " "
f = "Hello 123"
g = "1.234"
h = "Harry Potter And The Goblet Of Fire"


#isalnum - returns true if all character in the string are alphanumeric
print ( a,a.isalnum())
print (b,b.isalnum())
print (c,c.isalpha()) 
print (c,c.isalnum())
print (f,f.isalnum())
print (g,g.isalnum())

#isalpha - returns true if all character in the string are in the alphabet
print (a,a.isalpha())
print (b,b.isalpha())

#isdecimal - returns true if all character in the string are decimals

print (a,a.isdecimal())
print (g,g.isdecimal())
print (c,c.isdecimal())

#isdigit - returns true if all character in the string are digit
print (g,g.isdigit())
print (c,c.isdigit())
print (b,b.isdigit())

#isnumeric - returns true if all character in the string are numeric
print (b,b.isnumeric())
print (c,c.isnumeric())
print (e,e.isnumeric())

#islower - check if the string is lower case or not
print (a,a.islower())
print (d,d.islower())
print (b,b.islower())

#isupper - returns true if all character in the string are upper case
print (a,a.isupper())
print (b,b.isupper())
print (d,d.isupper())

#isspace - returns true if all character in the string are whitespaces
print (e,e.isspace())
print (f,f.isspace())

#istitle - returns true if the string follows the rules of a title
print (d,d.istitle())
print (f,f.istitle())
print (h,h.istitle())
```

```python
#endswith() -returns true if the string ends with the specified value
a = "Harry Potter"
print(a.endswith("r"))
print (a.endswith("t",6,9))


#startswith () -returns true if the string starts with the specified value
a = "Harry Potter"
print (a.startswith("H"))
print (a.startswith("P",6,9))

#swapcase() -swaps cases,lower case becomes upper case and vice versa
a = "Harry Potter"
print (a.swapcase())

#strip() - returns a trimmed version of the string
a = "  ***** Harry Potter *****  "
print (a)
print (a.strip(". ,*,"))

#spilt() - spilts the string at the specified separator, and retuns  a list
a ="#OOFD#BRB#OMW#TB"
b = "hello.my name is dear.i am 23 years old"
print (a.split("#"))
print (b.split("."))

#ljust () - returns a left justified version of the string
a = "Harry Potter"
x = a.ljust (20,"*")
print (x, "is my favourite movie")

#rjust () - returns a right justified version of the string
a = "Harry Potter"
x = a.rjust(20, "$")
print (x, "is my favourite movie")

#replace() - returns a string where a specvified value is replaced with a
# specifiied vallue
a = "my name is dear"
print (a.replace ("dear","janu"))

#rindex() - searches the string for a specified value and returns 
# the last position of where it was found
a = "Harry Potter and the Prisoner of Azkaban"
print (a.rindex("Prisoner"))
print (a.rindex("Harry"))


```
